```python
spam = ['apples', 'bananas', 'tofu', 'cats']
def sentence_format(spam):
    if len(spam) == 0:
        return ''
    elif len(spam) == 1:
        return spam[0]
    elif len(spam) == 2:
        return spam[0] + ' and ' + spam[1]
    else:
        return ', '.join(spam[:-1]) + ', and ' + spam[-1]

print(sentence_format(spam))
```
