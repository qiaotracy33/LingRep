```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
InternetSales = '/content/LingRep/data/InternetSales.csv'
# Specify the encoding to handle special characters
internet_sales = pd.read_csv(InternetSales, parse_dates=['BirthDate','DateFirstPurchase','OrderDate','DueDate','ShipDate'], delimiter=',', low_memory=False, index_col= False, encoding='latin1')
internet_sales["Year"] = pd.to_datetime(internet_sales["OrderDate"]).dt.year
internet_sales
```

```python
def catergory(name):
  if name.startswith(("Mountain-", "Touring", "Road-")):
    return "Bikes"

  elif any(word in name for word in ["Jersey", "Gloves", "Socks", "Vest", "Shorts", "Cap"]):
    return "Clothing"

  else:
    return "Accessories"

internet_sales["Product"] = internet_sales["EnglishProductName"].apply(catergory)
```

```python
sales_per_country = internet_sales.pivot_table(index='EnglishCountryRegionName', columns='Year', values='SalesAmount', aggfunc='sum', margins=True)
sales_per_country
```

```python
income= internet_sales.groupby('EnglishCountryRegionName')["YearlyIncome"].agg(['min','max','std'])
income
```

```python
gender= internet_sales.groupby(['EnglishCountryRegionName', 'Gender']).size()
gender
```

```python
catergory_sales = internet_sales.groupby("Product")['SalesAmount'].sum().sort_values(ascending=False)
top_sales = catergory_sales.index[0]
avg_price = internet_sales[internet_sales["Product"] == top_sales]["UnitPrice"].mean()
print(round(avg_price,2))
```
