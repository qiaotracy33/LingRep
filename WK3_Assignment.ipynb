### d. How can you prevent a program from crashing when it gets an error?
### We can use try-except statement.

```python
def collatz(number):
  if number % 2 == 0:
    result = number // 2
    print(result, end = ' ')
    return result
  elif number % 2 == 1:
    result = 3 * number + 1
    print(result, end = ' ')
    return result

def collatz_sequence(number):
  while number != 1:
    number = collatz(number)

def question(num):
  try:
    num = int(num)
    collatz_sequence(num)
  except ValueError:
    print('Please enter a number.')


num = input('Enter a number: ')
question(num)
```
