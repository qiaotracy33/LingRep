```python
import requests, os, time

TAG = 'cats'            # search by tag
MAX_DOWNLOADS = 10

os.makedirs('flickr', exist_ok=True)  # Store photos in ./flickr
num_downloads = 0

# TODO: Get the public feed for the tag.
feed_url = 'https://www.flickr.com/services/feeds/photos_public.gne'
params = {'tags': TAG, 'format': 'json', 'nojsoncallback': 1}
print(f'Getting Flickr feed for tag "{TAG}"...')
res = requests.get(feed_url, params=params)
res.raise_for_status()
items = res.json()['items']

for item in items:
    if num_downloads >= MAX_DOWNLOADS:
        break

    # TODO: Get the image URL (feed gives a small _m size; swap to a larger _z size).
    photo_URL = item['media']['m'].replace('_m.jpg', '_z.jpg')

    # TODO: Download the image.
    print(f'Downloading image {photo_URL}...')
    res = requests.get(photo_URL)
    if res.status_code != 200:      # skip photos that don't have this size
        print('  Skipped.')
        continue

    # TODO: Save the image to ./flickr.
    with open(os.path.join('flickr', os.path.basename(photo_URL)), 'wb') as image_file:
        for chunk in res.iter_content(100000):
            image_file.write(chunk)

    num_downloads += 1
    time.sleep(1)

print(f'Done. Downloaded {num_downloads} photos.')
```
