```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
apple_csv = '/content/drive/My Drive/MSSP607/Modules/Week05/aapl.csv'
amazon_csv = '/content/drive/My Drive/MSSP607/Modules/Week05/amzn.csv'
google_csv = '/content/drive/My Drive/MSSP607/Modules/Week05/goog.csv'
facebook_csv = '/content/drive/My Drive/MSSP607/Modules/Week05/fb.csv'
netflix_csv = '/content/drive/My Drive/MSSP607/Modules/Week05/nflx.csv'
apple_df = pd.read_csv(apple_csv)
amazon_df = pd.read_csv(amazon_csv)
google_df = pd.read_csv(google_csv)
facebook_df = pd.read_csv(facebook_csv)
netflix_df = pd.read_csv(netflix_csv)
```

```python
apple_df["ticker"] = "AAPL"
amazon_df["ticker"] = "AMZN"
google_df["ticker"] = "GOOG"
facebook_df["ticker"] = "FB"
netflix_df["ticker"] = "NFLIX"
```

```python
frames = []
frames.append(apple_df)
frames.append(amazon_df)
frames.append(google_df)
frames.append(facebook_df)
frames.append(netflix_df)

combined = pd.concat(frames, ignore_index=True)
combined.to_csv("faang.csv", index=False)         # e. 保存
combined.head()
```

```python
combined["date"] = pd.to_datetime(combined["date"])
combined["volume"] = combined["volume"].astype(int)
combined = combined.sort_values(by=["date", "ticker"])
combined.head()
```

```python
top7 = combined.nlargest(n=7, columns=["volume"])
top7
```

```python
melted = combined.melt(id_vars=["date", "ticker"], value_vars=["open", "high", "low", "close", "volume"])
melted.head()
```
