```python
summarize = lambda d: (
    d.groupby(['gender', 'test preparation course'])
     .agg(n=('total score', 'size'),
          mean_math=('math score', 'mean'),
          mean_reading=('reading score', 'mean'),
          mean_writing=('writing score', 'mean'),
          mean_total=('total score', 'mean'),
          high_prospect_rate=('admission prospect', lambda s: (s == 'high').mean()))
     .reset_index()
)

df_summary = summarize(df)
df_summary
```
