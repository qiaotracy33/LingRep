```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import geopandas as gpd
CA = pd.read_csv('/content/LingRep/data/CA_Legislative_Districts.csv', dtype={'DISTRICT':str}, encoding='utf-8-sig')
CA_geo = gpd.read_file('/content/LingRep/data/california-congressional-districts.geojson')
```

```python
CA_geo.plot(figsize=(6,8), facecolor="none", edgecolor="black", linewidth=0.5)
plt.title("California Legislative Districts")
plt.axis("off")
plt.show()
```

```python
CA_new = CA_geo.merge(CA, left_on='Name', right_on="DISTRICT")
CA_new['log_density'] = np.log10(CA_new.POPULATION / CA_new.AREA)
```

```python
CA_new.plot("POPULATION", cmap= "Accent", legend=True, edgecolor="black", linewidth=0.4, figsize=(6,8))
plt.title("California Legislative Districts")
plt.axis("off")
plt.show()
```

```python
CA_new.plot("log_density", cmap= "Accent", legend=True, edgecolor="black", linewidth=0.4, figsize=(6,8))
plt.title("California Population Density Heatmap")
plt.axis("off")
plt.show()
```
