```python
import pandas as pd
import numpy as np
from scipy import stats
df = pd.read_csv('/content/LingRep/data/Student Performance Data.csv')
df.head()
```

```python
male = df[df['gender'] == 'male']
female = df[df['gender'] == 'female']
a = male.loc[male["test preparation course"] == "completed", "total score"]
b = male.loc[male["test preparation course"] == "none", "total score"]
t, p = stats.ttest_ind(a, b, equal_var=False)

print(f"Prepared male's mean: {a.mean():.2f}，Unprepared male's mean: {b.mean():.2f}")
print(f"t = {t:.3f}, p = {p:.4f}")

if p < 0.05:
    print("Conclusion：p < 0.05，the difference is statistically significant。")
else:
    print("Conclusion：p >= 0.05，the difference is not statistically significant。")
```
